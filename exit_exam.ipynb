# Task 2 - Data Preprocessing

```python
df = df.drop(columns=["Response_ID", "Timestamp"], errors="ignore")
#we can remove these columns because it has unique values and not necessary for our model
```

```python
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
```

```python
df = df.replace(["", " ", "nan", "NaN", "None", "NULL"], np.nan)
```

```python
print("\nMissing values:\n", df.isnull().sum().sort_values(ascending=False))
```

```python
for col in df.columns:
    if df[col].dtype in ["float64", "int64"]:
        if df[col].isnull().any():
            df[col] = df[col].fillna(df[col].median())

# Fill object/string columns with mode
for col in df.columns:
    if df[col].dtype == "object":
        if df[col].isnull().any():
            mode_val = df[col].mode(dropna=True)
            if not mode_val.empty:
                df[col] = df[col].fillna(mode_val.iloc[0])
            else:
                df[col] = df[col].fillna("Unknown")
```

```python
# Define mapping dictionaries
income_map = {
    'Under $1,000': 500,
    '$1,000-3,000': 2000,
    '$3,000-5,000': 4000,
    '$5,000-10,000': 7500,
    '$10,000+': 12000
}

spend_map = {
    'Under $50': 25,
    '$50-150': 100,
    '$150-300': 225,
    '$300-500': 400,
    '$500+': 600
}

freq_map = {
    'Never': 0,
    'Rarely': 1,
    'Occasionally': 2,
    'Often': 3,
    'Always': 4
}

# Apply mappings to create new numerical columns
df['Monthly_Income_Value'] = df['Monthly_Income_Range'].map(income_map)
df['Spend_Value'] = df['Avg_Monthly_Spend_NonEssentials'].map(spend_map)
df['Online_Shopping_Frequency_Value'] = df['Online_Shopping_Frequency'].map(freq_map)

# Drop the original categorical columns as they are now represented numerically
df = df.drop(columns=['Monthly_Income_Range', 'Avg_Monthly_Spend_NonEssentials', 'Online_Shopping_Frequency'], errors='ignore')

print("Categorical range columns mapped to numerical values and original columns dropped.")
print(df.head())
```

```python
# Identify numerical features (including the newly created _Value columns)
numerical_features = df.select_dtypes(include=['number', 'float64', 'int64']).columns.tolist()
# Exclude 'Response_ID' if it was not already dropped and is still in df
if 'Response_ID' in numerical_features:
    numerical_features.remove('Response_ID')


# Identify categorical features (remaining object columns)
categorical_features = df.select_dtypes(include='object').columns.tolist()
# Remove 'Best_Shopping_Mode' as it is the target variable
if 'Best_Shopping_Mode' in categorical_features:
    categorical_features.remove('Best_Shopping_Mode')

print("Updated Numerical Features:", numerical_features)
print("Updated Categorical Features:", categorical_features)

numerical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])

# Create a column transformer to apply different transformations to different columns
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numerical_transformer, numerical_features),
        ('cat', categorical_transformer, categorical_features)
    ],
    remainder='passthrough' # Keep any other columns not specified (e.g., if there are any that slipped through)
)

print("Preprocessor redefined with updated feature lists.")
```

```python
print("Number of duplicate rows:", df.duplicated().sum())
df = df.drop_duplicates().copy()
print("Dataset shape after removing duplicates:", df.shape)
```

```python
#Splitting the target variable from the dataset and creating X,y
X = df.drop(columns=['Best_Shopping_Mode'])
y = df['Best_Shopping_Mode']
```

```python
# The manual categorical encoding has been moved to the preprocessor pipeline.
# X = pd.get_dummies(X, drop_first=False)
# X.head()
print("Manual categorical encoding removed, handled by preprocessor.")
```

```python
#Encoding the target variable using map()
from sklearn.preprocessing import LabelEncoder
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y_encoded,test_size=0.2,random_state=42,stratify=y_encoded)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

#Machine Learning Model

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report, confusion_matrix
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

logistic_model = Pipeline([
    ("preprocessor", preprocessor), # Use the preprocessor
    ("model", LogisticRegression(
        max_iter=5000,
        solver="lbfgs",
        multi_class="auto",
        random_state=42 # Added random_state for reproducibility
    ))
])

logistic_model.fit(X_train, y_train)
y_pred = logistic_model.predict(X_test)

```

### Model Training and Evaluation



```python
models = {
    "Logistic Regression": LogisticRegression(max_iter=5000, random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=200, random_state=42),
    "Gradient Boosting": GradientBoostingClassifier(random_state=42)
}

results = []

for name, model in models.items():
    pipe = Pipeline([
        ("preprocessor", preprocessor),
        ("model", model)
    ])

    print(f"\nTraining {name}...")
    pipe.fit(X_train, y_train)
    y_pred = pipe.predict(X_test)

    acc = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, average="weighted", zero_division=0)
    recall = recall_score(y_test, y_pred, average="weighted", zero_division=0)
    f1 = f1_score(y_test, y_pred, average="weighted", zero_division=0)

    print(f"\n===== {name} =====")
    print("Accuracy :", round(acc, 4))
    print("Precision:", round(precision, 4))
    print("Recall   :", round(recall, 4))
    print("F1-score :", round(f1, 4))

    print(classification_report(y_test, y_pred, target_names=label_encoder.classes_))

    cm = confusion_matrix(y_test, y_pred)
    plt.figure(figsize=(8, 6))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
                xticklabels=label_encoder.classes_,
                yticklabels=label_encoder.classes_)
    plt.title(f"{name} Confusion Matrix")
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.show()

    results.append({
        "Model": name,
        "Accuracy": acc,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "Trained_Model": pipe # Store the trained pipeline
    })


results_df = pd.DataFrame(results)
print("\nFinal comparison table:")
display(results_df.sort_values(by="Accuracy", ascending=False).reset_index(drop=True))
```

### Saving the Best Model

Based on the evaluation metrics, we will identify the best performing model and save it as a pickle file. This allows us to reuse the trained model without retraining.

```python
import pickle

# Get the best model based on accuracy
best_model_row = results_df.sort_values(by="Accuracy", ascending=False).iloc[0]
best_model_name = best_model_row["Model"]
best_model_pipeline = best_model_row["Trained_Model"]

# Define the filename for the pickle file
pickle_filename = f"{best_model_name.replace(' ', '_').lower()}_model.pkl"

# Save the best model pipeline to a pickle file
with open(pickle_filename, 'wb') as file:
    pickle.dump(best_model_pipeline, file)

print(f"The best model ({best_model_name}) has been saved to {pickle_filename}")
```
